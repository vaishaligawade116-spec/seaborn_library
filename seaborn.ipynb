```python
#importing the required libraries
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
#getting the built-in datasets of seaborn
sns.get_dataset_names()
```

```python
#loading dataset
df=sns.load_dataset("iris")
```

```python
df
```

```python
#regression-continues data
#two types
#binary class-only two classifications(positive ,negative)
#multi class-more than two classifications(vanila,strawberry,chocolate)
```

```python
df.species.unique() #used to find all the unique (distinct) values in the species column of a DataFrame.
```

```python
df.species.value_counts() #used to count how many times each unique value appears in the species column.
```

```python
df.head() #used to display the first 5 rows of a DataFrame by default.
```

```python
df.tail() #Displays the last 5 rows
```

```python
#lineplot() to create a line chart showing the relationship between sepal_length and sepal_width.
#ci=None — in older Seaborn versions, disables the confidence interval shading.
sns.lineplot(x='sepal_length',y='sepal_width',data=df,ci=None)#confidence interval
plt.show() #displays the graph.
```

```python
#This code creates a Seaborn line plot with different colored lines for each species in the Iris dataset.
#hue='species' — assigns different colors to different species.
sns.lineplot(x='sepal_length',y='sepal_width',data=df,hue='species',ci=None)
plt.show()
```

```python
sns.lineplot(x='sepal_width',y='species',data=df,hue='species',ci=None)
plt.show()
```

```python
sns.lineplot(x='petal_length',y='species',data=df,hue='species',ci=None)
plt.show()
```

```python
#scatter plot
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue='species')
plt.title("Scatter plot")
plt.show()
```

```python
sns.scatterplot(x='petal_length',y='petal_width',data=df,hue='species')
plt.title("Scatter plot")
plt.show()
```

```python
#barplot-A bar plot is a chart used to compare a numerical value across different categories.
sns.barplot(x='species',y='sepal_length',data=df)
plt.show()
```

```python
#count plot value_count()
#count plot basically counts the categories and return a count of their occurence
sns.countplot(x='species',data=df)
plt.show()
```

```python
#Boxplot
#A boxplot is a graph used to understand the distribution of numerical data. It helps us identify the median, quartiles, spread of data, and outliers.
#a boxplot is sometimes known as the box and whisker plot,it shows the distributin of quantitative data
##it represents the comparisons between variables
sns.boxplot(x='sepal_width',data=df)
plt.show()
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
#violinplot
#A violin plot is a graph used to understand the distribution of numerical data.
#It combines a boxplot with a density plot to show where data values are concentrated and how the data is distributed.
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
#striplot-it basically creates a scatter plot based on the category
sns.stripplot(x='species',y='petal_width',data=df)
plt.show()
```

```python
#histogram
#A histogram is a graph used to show the frequency distribution of numerical data.
#It tells us how many observations fall within different ranges of values.
sns.histplot(x='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width',data=df,hue='species',kde=True)#kernel density estimation
plt.show()
#kde is approximation of histogram to density curve
```

```python
import warnings
warnings.filterwarnings('ignore')
#distplot
#distplot is basically used for univariate set of observations
sns.distplot(df['petal_width'],color="g")#hist=true,kde=false
plt.show()
```

```python
sns.distplot(df['petal_width'],color="g",hist=True,kde=False)
plt.show()
```

```python
#pairplot-
#A pairplot is used to visualize relationships between multiple numerical variables in a dataset. 
 #It creates a collection of graphs showing how variables relate to one another.
sns.pairplot(data=df,hue='species')
plt.show()
```

```python
#categorical plot
#types of catplot-violin,swam
sns.catplot(x='sepal_length',data=df,kind='violin')
plt.show()
```

```python
#categorical plot
sns.catplot(x='species',data=df,kind='violin')
plt.show()
```

```python
#categorical plot
sns.catplot(x='species',y='sepal_length',data=df,kind='swarm') #Shows individual observations without overlapping points as much as possible.
plt.show()
```

```python
#heatmap
#A heatmap is a graph that uses different colors to represent numerical values. 
#It helps us quickly identify high values, low values, and patterns in data.
tc=df.corr(numeric_only=True)
tc
```

```python
sns.heatmap(tc,annot=True)
plt.show()
```
